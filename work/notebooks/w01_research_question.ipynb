## 2. The question: decision, action, cost of a wrong call

Search question: Which content pages are declining and worth prioritizing for 
review, based on signals we can observe before the outcome happens?

Unit of analysis: A single content page.

Output: A ranked list of pages, ordered by how likely they are to be declining 
and worth reviewing first.

The decision this supports: A content team has limited time — they can't review 
every page every week. This ranking tells them which pages to look at first.

The action someone takes: A content editor opens the top-ranked pages and 
decides whether to refresh, rewrite, or leave them as-is.

Cost of a wrong recommendation: A false positive (flagged as declining but 
actually fine) wastes an editor's time reviewing a healthy page. A false 
negative (a genuinely declining page gets missed) is more expensive, since 
the page keeps losing traffic unnoticed and the problem compounds over time.

Why data/ML can help: With thousands of pages, no team can manually review 
all of them regularly. A ranking model turns an unmanageable list into a 
short, prioritized one.
